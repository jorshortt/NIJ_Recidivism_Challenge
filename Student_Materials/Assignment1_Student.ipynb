<h2>Background</h2>
Two of the major reasons for the existence of American prisons are usually cited as rehabilitation and deterrence (MacKENZIE, 2001). Essentially, two of the main functions of prisons are to: 

1) Rehabilitate criminals and guide them into lives as productive members of society, and 
2) Deter citizens from committing crimes out of fear of punishment.

If prison systems are accomplishing these goals, then it should, in theory, be evidenced by low **recidivism**.

The National Institute of Justice has contracted your help to investigate **recidivism data** to determine if our prisons are working the way they were intended.
